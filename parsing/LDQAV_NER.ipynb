# Named-entity recognition, NER

+ Named-entities: names, locations, organizations, dates...
+ Needed for reference, metonymy, coreference, literary analysis, summarization etc.
+ Difficulties:
    - homonymy
    - technical: tags, boundaries
+ BIOES:

  - B – beginning – the first token;
  - I – inside – inside the span;
  - О – outside – not part of the named-entity;
  - E – ending – the last token;
  - S – single – single token entity.

## Stanza

+ https://stanfordnlp.github.io/stanza/
+ NLP library
+ more than 70 languages, including Russian
+ tokenization, lemmatization, morphological and syntactic parsing, NER

```python
!pip install stanza --q
```

```python
import stanza
stanza.download('en')
```

To do NER, we need to tokenize the text first.

```python
def stanza_nlp(text):
  nlp = stanza.Pipeline(lang='en', processors='tokenize,ner')
  doc = nlp(text)
  print(*[f'entity: {ent.text}\ttype: {ent.type}' for sent in doc.sentences for ent in sent.ents], sep='\n')
```

```python
stanza_nlp(eng_text)
```

If BIOES NER tags are needed for each token:

```python
nlp = stanza.Pipeline(lang='en', processors='tokenize,ner')
doc = nlp(eng_text)
print(*[f'token: {token.text}\tner: {token.ner}' for sent in doc.sentences for token in sent.tokens][10:50], sep='\n')
```

## SpaCy

+ NLP Library
+ a number of languages
+ About spaCy: https://spacy.io/usage


```python
!pip install -U pip setuptools wheel
!pip install -U spacy
!python -m spacy download en_core_web_sm
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp(eng_text)

for ent in doc.ents:
    print(ent.text, ent.start_char, ent.end_char, ent.label_)
```

SpaCy offers 18 tags (the list can be found [here](https://towardsdatascience.com/named-entity-recognition-ner-using-spacy-nlp-part-4-28da2ece57c6)), but you can also add your own:

```python
from spacy.tokens import Span

nlp = spacy.load("en_core_web_sm")
doc = nlp(eng_text)
ents = [(e.text, e.start_char, e.end_char, e.label_) for e in doc.ents]
print('Before', ents)
# the model didn't recognize "Nord Stream" as an entity :(

ns2_ent = Span(doc, 16, 18, label="OBJ") # create a Span for the new entity, OBJ = object
doc.ents = list(doc.ents) + [ns2_ent]

ents = [(e.text, e.start_char, e.end_char, e.label_) for e in doc.ents]
print('After', ents)
#('Nord Stream', 82, 93, 'OBJ') got recognized as an object entity
```

```python
len([ent for ent in doc.ents if ent.label_=='OBJ'])
```

```python
from spacy import displacy
```

```python
displacy.render(doc,style = "ent",jupyter=True)
```
