# Classes

Every object is a representative of a certain class.

Class describes the attributes and the methods of the object. Attributes are variables that hold data (like name or breed), while methods are actions the object can perform (like bark()).

```__init__``` -- this special method runs automatically whenever you create a new object from the class. It sets up the initial values (attributes) for your object.

```self``` -- this keyword refers to the specific object you are currently creating or manipulating. It must be the first parameter of any instance method inside the class.

Classes are usually upper-cased, while the objects lower-cased.

````python
class Cat:

    number_of_eyes = 2

    def __init__(self, eye_color, fur_color):
        self.eye_color = eye_color
        self.fur_color = fur_color

    def talk(self):
        return "Meow!"
````

````python
princess = Cat(eye_color="green", fur_color="grey")
vasya = Cat(eye_color="yellow", fur_color="red")
````

## Inheritance

Classes allow you to inherit traits and functionality from other classes, preventing code repetition.

````python
class Animal:

    def __init__(self, name, age, eye_color, fur_color):
        self.name = name
        self.age = age
        self.eye_color = eye_color
        self.fur_color = fur_color

    @staticmethod # we are not using any attributes, e.g. self.name
    def talk():
        return "..." # a string

    def birthday(self):
        self.age += 1
        return f"{self.name} is already {self.age}!"

    def __str__(self):
        return f"Name {self.name}.\nAge {self.age}.\nEyes: {self.eye_color}.\nFur: {self.fur_color}"
````

````python
class Dog(Animal):

    @staticmethod
    def talk():
        return "Гав!"

    @staticmethod
    def get_stick():
        return "Гав!Гав!"

class Cat(Animal):

    @staticmethod
    def talk():
        return "Мяу!"
````

````python
princess = Cat("Princess", 3, "green", "grey")
ron = Cat("Ron", 5, "yellow", "red")

sirius = Dog("Sirius", 4, "black", "black")
````

````python
sirius.get_stick()
````

````python
princess.get_stick()
````
